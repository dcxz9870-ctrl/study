> 우선순위는 `**` > `* / // %` > `+ -` > 비교 > `not` > `and` > `or`

### 문제 5-1 · Lv.1

<details>
<summary>🔍 <b>힌트</b></summary>

`//` 는 몫, `%` 는 나머지, `**` 는 거듭제곱입니다.

</details>

<details>
<summary>📤 <b>출력 예시</b></summary>

```
a + b  = 22

**해설:** `/` 는 나누어떨어져도 **결과가 항상 실수**입니다 (`10/2` → `5.0`). 정수 몫이 필요하면 `//`. `%`(나머지)는 **짝/홀 판별**(`n % 2 == 0`)이나 배수 판별에 정말 자주 씁니다.

<details>

컴프리헨션 기본형은 `[식 for 변수 in 반복가능 if 조건]`. 필터는 if 를 뒤에, 값 선택은 if 를 앞에 씁니다.

</details>

<details>
<summary>📤 <b>출력 예시</b></summary>

```
[6, 16, 4, 30, 14, 40]
[15, 20]
['홀', '짝', '짝', '홀', '홀', '짝']
1 3
2 8
3 2
4 15
5 7
6 20
A 90

C 70

</details>

<details>
<summary>💡 <b>정답 보기</b></summary>

```python
nums = [3, 8, 2, 15, 7, 20]
names = ["A", "B", "C"]
scores = [90, 80, 70]

print([n * 2 for n in nums])                        # 1. 변환
print([n for n in nums if n > 10])                  # 2. 필터
print(["짝" if n % 2 == 0 else "홀" for n in nums])  # 3. 조건부 값

# enumerate — 번호와 값을 함께
for i, n in enumerate(nums, start=1):
    print(i, n)

# zip — 두 리스트를 짝지어서
for name, score in zip(names, scores):
    print(name, score)
```

**해설 (if 위치가 다릅니다!):**
| 목적 | 형태 |
|---|---|
| 필터 | `[n for n in nums **if 조건**]` — if 가 **뒤** |
| 조건부 값 | `[**A if 조건 else B** for n in nums]` — if 가 **앞** |

`enumerate(리스트, start=1)` 로 시작 번호를 바꿀 수 있고, `zip` 은 **짧은 쪽 길이에 맞춰** 멈춥니다.

</details>

`return max(nums), min(nums), ...` 처럼 쉼표로 나열하면 튜플로 반환됩니다. 2번은 `global` 키워드가 필요합니다.

def factorial(n):
    if n <= 1:                       # 종료 조건 (base case)
        return 1
    return n * factorial(n - 1)      # 자기 자신을 호출

def fib(n):
    if n <= 1:

    return fib(n - 1) + fib(n - 2)

print(factorial(5))

# 📌 시험 직전 최종 체크리스트

### 파이썬 기본
| 헷갈리기 쉬운 것 | 정리 |
|---|---|
| `/` vs `//` | `/` 는 **항상 실수**(`10/2` → `5.0`), `//` 는 몫 |
| `int(3.9)` | `3` — **반올림 아니고 버림** |
| `bool("False")` | **`True`** (빈 문자열이 아니므로) |
| `range(1, 5)` | `1,2,3,4` — **끝 미포함** |
| `sort()` vs `sorted()` | 원본 정렬 후 `None` 반환 vs 새 리스트 반환 |
| `remove()` vs `pop()` | **값**으로 삭제 vs **인덱스**로 꺼내기 |
| `find()` vs `index()` | 없으면 `-1` vs 없으면 **에러** |
| `"구분자".join(리스트)` | join 은 **구분자가 앞** |
| `==` vs `is` | **값** 비교 vs **객체** 비교 |
| `[[0]*3]*3` | 같은 리스트 참조 → **컴프리헨션 사용** |
| `*args` / `**kwargs` | 튜플로 받음 / 딕셔너리로 받음 |
| 전역변수 수정 | 함수 안에서 대입하려면 **`global`** 필요 |
| 조건문 순서 | **좁은 조건(15의 배수, 90점)부터** 먼저 |

### Pandas
| 헷갈리기 쉬운 것 | 정리 |
|---|---|
| `df["A"]` vs `df[["A"]]` | **Series** vs **DataFrame** |
| `loc` vs `iloc` | 라벨(**끝 포함**) vs 위치(**끝 미포함**) |
| 조건 2개 이상 | `and`/`or` ❌ → **`&`, `|` + 괄호** ✅ |
| `axis=0` vs `axis=1` | 세로↓(열 단위) vs **가로→(행 단위)** |
| `fillna`, `dropna`, `drop` | **원본을 바꾸지 않음** → 다시 대입 필요 |
| 통계 함수와 `NaN` | 기본적으로 **NaN 을 빼고** 계산 |
| `shape`, `dtypes`, `values` | **괄호 없는 속성** (`df.shape()` ❌) |
| 문자열로 읽힌 숫자 | `astype(int)` 로 변환해야 집계 가능 |
| groupby 기본형 | `df.groupby("기준")["대상"].집계()` |

**화이팅! 🔥**

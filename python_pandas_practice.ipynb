```python
import pandas as pd

name = pd.Series(["A", "B", "C"], name="이름")
```

### 문제 11-3 · Lv.3

DataFrame 에 결측치가 섞여 있습니다.

```python
df = pd.DataFrame({
    "이름": ["A", "B", "C", "D", "E"],
    "국어": [90, np.nan, 70, 85, np.nan],
    "영어": [80, 75, np.nan, 95, 60],
})
```

1. 열별 결측치 개수를 구하세요.
2. 각 열의 결측치를 **그 열의 평균**으로 채우세요.
3. `총점`, `평균` 열을 추가하고 평균 기준 내림차순 정렬하세요.

<details>
<summary>💡 <b>정답 보기</b> (클릭)</summary>

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "이름": ["A", "B", "C", "D", "E"],
    "국어": [90, np.nan, 70, 85, np.nan],
    "영어": [80, 75, np.nan, 95, 60],
})

print(df.isna().sum())        # 열별 결측치 개수

num_cols = ["국어", "영어"]
df[num_cols] = df[num_cols].fillna(df[num_cols].mean())
print(df)

df["총점"] = df[num_cols].sum(axis=1)
df["평균"] = df[num_cols].mean(axis=1).round(1)

print(df.sort_values("평균", ascending=False))
```

**해설:**
- `df.isna().sum()` 은 **열별** 결측치 개수를 한 번에 보여줍니다.
- `fillna(df.mean())` 은 열마다 각자의 평균으로 채워줍니다.
- **`axis=1` 은 행 방향(가로) 계산**, `axis=0` 은 열 방향(세로) 계산입니다. 총점은 가로로 더해야 하므로 `axis=1`.

</details>

## 📌 시험 직전 체크리스트

| 헷갈리기 쉬운 것 | 정리 |
|---|---|
| `/` vs `//` | `/` 는 항상 실수, `//` 는 몫 |
| `sort()` vs `sorted()` | 앞은 원본 변경 후 `None` 반환, 뒤는 새 리스트 반환 |
| `remove()` vs `pop()` | 값으로 삭제 vs 인덱스로 꺼내기 |
| `==` vs `is` | 값 비교 vs 객체 비교 |
| `bool("False")` | **True** (빈 문자열이 아니므로) |
| `range(1, 5)` | 1,2,3,4 — **끝 미포함** |
| `[[0]*3]*3` | 같은 리스트 참조 → 컴프리헨션 사용 |
| `df["A"]` vs `df[["A"]]` | Series vs DataFrame |
| `loc` vs `iloc` | 라벨(끝 포함) vs 위치(끝 미포함) |
| DataFrame 조건 결합 | `and/or` ❌ → `&`, `|` 와 괄호 ✅ |
| `axis=0` vs `axis=1` | 열 방향(세로) vs 행 방향(가로) |
